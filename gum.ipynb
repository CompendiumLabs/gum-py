```python
import gum
import numpy as np
import pandas as pd
from gum import G, C, px, em
```

```python
np.random.seed(1337)
```

```python
G.Box(
    G.HStack(
        G.Square(width=px(100), border_radius=px(8), fill=C.blue),
        G.Square(width=px(100), border_radius=px(8), fill=C.red),
        gap=em(1),
    ),
    padding=em(1),
)
```

```python
G.Frame(G.Text('Hello, world!'), border_radius=px(8), padding=em(1))
```

```python
path1 = G.SymLine(fy=C.sin, xlim=(0, 2*C.pi), stroke=C.blue, stroke_width=px(2))
path2 = G.SymLine(fy=C.cos, xlim=(0, 2*C.pi), stroke=C.red, stroke_width=px(2))
G.VStack(
    G.Plot(path1, ylim=(-1.5, 1.5), width=px(600), aspect=2, grid=True),
    G.Plot(path2, ylim=(-1.5, 1.5), width=px(600), aspect=2, grid=True),
    gap=em(1),
)
```
